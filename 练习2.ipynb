```python
from typing import List

def plusOne(digits: List[int]) -> List[int]:
    # 提示 1：像做竖式加法一样，从数组最后一位开始
    for i in range(len(digits) - 1, -1, -1):
        
        # 提示 2：当前位小于 9 时加一后即可立即返回
        if digits[i] < 9:
            digits[i] += 1
            return digits
        
        # 提示 3：当前位等于 9 时把它改成 0
        digits[i] = 0
        
    # 提示 3（续）：若所有位都是 9，需要在最前面补 1
    # 此时 digits 全部变成了 0，例如 [9,9,9] 变成了 [0,0,0]
    # 在最前面插入 1，变成 [1,0,0,0]
    return [1] + digits 
```

```python
print(plusOne([1, 2, 3]))  # 输出: [1, 2, 4]
print(plusOne([1, 2, 9]))  # 输出: [1, 3, 0]
print(plusOne([9, 9, 9]))  # 输出: [1, 0, 0, 0]
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverseList(head: ListNode) -> ListNode:
    # 初始化两个指针
    # prev 指向已反转部分的头节点，初始为空
    # cur 指向当前正在处理的节点，初始为原链表的头节点
    prev = None
    cur = head
    
    # 遍历整个链表
    while cur:
        # 提示 2：改写 cur.next 前，必须先用 nxt 保存原来的下一个节点
        nxt = cur.next  
        
        # 核心操作：反转指针方向
        cur.next = prev
        
        # 提示 3：反转 cur.next 后，令 prev = cur，再令 cur = nxt
        # 将 prev 和 cur 指针向前移动
        prev = cur
        cur = nxt
        
    # 循环结束时，cur 指向 None，prev 指向反转后的新头节点
    return prev
```

```python
if __name__ == "__main__":
    # 构建 1 -> 2 -> 3
    node3 = ListNode(3)
    node2 = ListNode(2, node3)
    node1 = ListNode(1, node2)
    
    new_head = reverseList(node1)
    
    # 打印反转后的链表
    while new_head:
        print(new_head.val, end=" -> ")
        new_head = new_head.next
```
