```python
def canConstruct(s: str, t: str) -> bool:
    # 提示 1：两个字符串长度不同，它们的字符组成一定不同
    if len(s) != len(t):
        return False
    
    # 初始化哈希表（用字典记录字符出现的次数）
    count = {}
    
    # 提示 2：用哈希表记录每种字母的数量；扫描 s 时把对应计数加 1
    for char in s:
        count[char] = count.get(char, 0) + 1
        
    # 提示 3：扫描 t 时把对应计数减 1
    for char in t:
        # 如果 t 中出现了 s 里没有的字符，或者某个字符数量超过了 s 中的数量
        if char not in count or count[char] == 0:
            return False
        count[char] -= 1
        
    # 因为长度已经相等，并且每个字符数量都抵消了，说明字符组成相同
    return True

# 测试用例
if __name__ == "__main__":
    print(canConstruct("anagram", "nagaram"))  # 输出 True
    print(canConstruct("rat", "car"))          # 输出 False
```
