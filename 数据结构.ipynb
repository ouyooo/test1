```python
class ListNode:
    """链表节点类"""
    def __init__(self, val: int):
        self.val: int = val  # 节点值
        self.next: ListNode | None = None  # 后继节点引用

# 使用多种基本数据类型来初始化数组
numbers: list[int] = [0] * 5
decimals: list[float] = [0.0] * 5
# Python 的字符实际上是长度为 1 的字符串
characters: list[str] = ['0'] * 5
bools: list[bool] = [False] * 5
# Python 的列表可以自由存储各种基本数据类型和对象引用
data = [0, 0.0, 'a', False, ListNode(0)]
```

```python
"""练习"""
def hammingWeight(n: int) -> int:
    count = 0
    # 当 n 不为 0 时循环
    while n > 0:
       #n & (n - 1) ： n 中最右边的一个 1 变成 0
        n = n & (n - 1)
        # 每消除一个 1，计数器加一
        count += 1
        
    return count
```

```python
#测试

if __name__ == "__main__":
    print(hammingWeight(11))  # 11 的二进制是 1011，有 3 个 1，输出 3
    print(hammingWeight(128)) # 128 的二进制是 10000000，有 1 个 1，输出 1
    print(hammingWeight(0))   # 0 的二进制是 0，输出 0
```
